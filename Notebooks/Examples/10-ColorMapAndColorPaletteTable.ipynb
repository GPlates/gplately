### This example demonstrates how to work with Matplotlib colormap and GMT Color Palette Table(cpt).


```python
import warnings
import cartopy.crs as ccrs
import matplotlib.pyplot as plt
from gplately import Raster
from gplately.plot import get_age_grid_cmap
from plate_model_manager import PlateModelManager

warnings.filterwarnings("ignore", category=UserWarning, module="gplately")

pmm_model = PlateModelManager().get_model("Zahirovic2022", data_dir="plate-model-repo")

agegrid_raster = Raster(data=pmm_model.get_raster("AgeGrids", 0))

fig = plt.figure(figsize=(8, 4))
ax1 = fig.add_subplot(111, projection=ccrs.Mollweide(0))

# Here is how to use the built-in colormap for the agegrid raster.
im = agegrid_raster.plot(ax1, cmap=get_age_grid_cmap(), vmin=0, vmax=200)

if im:
    fig.colorbar(im, orientation="horizontal", shrink=0.4, pad=0.05, label="Age (Ma)")

plt.show()
```

```python
# Here is how to access the age_grid.cpt file using importlib.resources
# And use the .cpt file directly in PyGMT plotting.
from importlib.resources import files
from gplately.auxiliary import get_pygmt_basemap_figure

cpt_file = files("gplately").joinpath("data", "agegrid.cpt")
print(cpt_file)

fig = get_pygmt_basemap_figure(
    projection="N180/10c", region="d", frame=["xafg30", "yafg30"]
)

agegrid_raster.plot(
    fig,
    cmap=str(cpt_file),
    nan_transparent=True,
    use_gmt=True,
)

fig.show(width=1200, crop="+m0.4c")
```

```python
# Here is how to turn a GMT .cpt file into a matplotlib colormap object.
from gplately.plot import get_cmap_from_gmt_cpt
import numpy as np

plt.figure(figsize=(10, 1))
plt.imshow(
    [np.linspace(0, 200, 256)],
    aspect="auto",
    cmap=get_cmap_from_gmt_cpt(cpt_file),
    vmin=0,
    vmax=200,
)
plt.xticks([])
plt.yticks([])
plt.colorbar(orientation="horizontal", label="Seafloor Age (Ma)", shrink=0.8, pad=0.2)
plt.title("Colormap Created from GMT CPT")
plt.show()
```
